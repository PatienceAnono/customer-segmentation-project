## 6. Data Cleaning

> **Principle:** Every cleaning decision is documented and justified. We never silently drop data without a clear reason. All changes are logged.

````python
#Work on a clean copy — never modify raw data
df = df_raw.copy()
cleaning_log = []  # Track every decision

# Step 1: Drop duplicates
before = len(df)
df = df.drop_duplicates()
after = len(df)
cleaning_log.append({'Step': '1. Drop duplicates', 'Rows removed': before - after, 'Notes': 'Exact duplicate rows removed'})
print(f'Step 1 — Duplicates removed: {before - after} rows  ({before} → {after})')
````

````python
#  Step 2: Standardise gender 
gender_map = {'M': 'Male', 'F': 'Female', 'Male': 'Male', 'Female': 'Female'}
df['gender'] = df['gender'].map(gender_map)
# Remaining NaN → 'Unknown'
df['gender'] = df['gender'].fillna('Unknown')

cleaning_log.append({'Step': '2. Standardise gender', 'Rows removed': 0,
                     'Notes': 'M→Male, F→Female, NaN→Unknown'})
print(f'Step 2 — Gender values after standardisation:')
print(df['gender'].value_counts())
````

````python
#Step 3: Standardise preferred_category 
df['preferred_category'] = df['preferred_category'].str.title()

cleaning_log.append({'Step': '3. Standardise preferred_category', 'Rows removed': 0,
                     'Notes': 'Converted all values to Title Case'})
print('Step 3 — Categories after standardisation:')
print(df['preferred_category'].value_counts())
````

````python
#Step 4: Parse date columns
df['signup_date']        = pd.to_datetime(df['signup_date'])
df['last_purchase_date'] = pd.to_datetime(df['last_purchase_date'])

# Derive customer tenure in days
ANALYSIS_DATE = pd.Timestamp('2025-05-01')  # Fixed reference date for reproducibility
df['tenure_days'] = (ANALYSIS_DATE - df['signup_date']).dt.days

cleaning_log.append({'Step': '4. Parse dates + derive tenure', 'Rows removed': 0,
                     'Notes': f'Reference date: {ANALYSIS_DATE.date()}'})
print('Step 4 — Date columns parsed. Tenure (days) sample:')
print(df[['customer_id', 'signup_date', 'last_purchase_date', 'tenure_days']].head(4))
````

````python
# Step 5: Impute missing numeric values
# age → median (robust to outliers)
age_median = df['age'].median()
df['age'] = df['age'].fillna(age_median)
df['age'] = df['age'].round(0).astype(int)

# frequency → median
freq_median = df['frequency'].median()
df['frequency'] = df['frequency'].fillna(freq_median)

cleaning_log.append({'Step': '5. Impute missing numerics', 'Rows removed': 0,
                     'Notes': f'age → median ({age_median:.0f}), frequency → median ({freq_median})'})
print(f'Step 5 — Imputed: age median = {age_median:.0f} | frequency median = {freq_median}')
print(f'Remaining nulls: {df.isnull().sum().sum()}')
````

````python
#Step 6: Handle outliers in monetary_value 
# Method: Cap at 99th percentile (Winsorisation)
# Rationale: Extreme spenders are real customers — we clip, not drop
p99 = df['monetary_value'].quantile(0.99)
outlier_count = (df['monetary_value'] > p99).sum()

df['monetary_value_raw'] = df['monetary_value'].copy()  # Keep original for reference
df['monetary_value']     = df['monetary_value'].clip(upper=p99)

cleaning_log.append({'Step': '6. Cap monetary_value outliers', 'Rows removed': 0,
                     'Notes': f'Winsorised at 99th percentile = ${p99:.2f} | {outlier_count} values capped'})
print(f'Step 6 — monetary_value capped at 99th pct: ${p99:.2f}')
print(f'         {outlier_count} values were capped')
````

````python
#Step 7: Reset index
df = df.reset_index(drop=True)
cleaning_log.append({'Step': '7. Reset index', 'Rows removed': 0, 'Notes': 'Clean index after deduplication'})

#Cleaning Log Summary 
print('\n' + '━' * 60)
print('  DATA CLEANING LOG')
print('━' * 60)
display(pd.DataFrame(cleaning_log))
print(f'\nFinal clean dataset: {df.shape[0]:,} rows × {df.shape[1]} columns')
print(f'Remaining null values: {df.isnull().sum().sum()}')
````

## 8. RFM Feature Engineering

RFM stands for **Recency**, **Frequency** and **Monetary Value** — three behavioural dimensions that together describe a customer's relationship with the business.

| Dimension | Question | Higher is better? |
|---|---|---|
| **Recency** | How recently did the customer buy? | Lower days = better |
| **Frequency** | How often do they buy? | Higher = better |
| **Monetary** | How much do they spend? | Higher = better |

````python
#Extract RFM columns 
rfm = df[['customer_id', 'recency_days', 'frequency', 'monetary_value']].copy()

# Rename for clarity
rfm.columns = ['customer_id', 'R', 'F', 'M']

print('RFM Table — first 8 rows')
display(rfm.head(8))

print('\nRFM Summary Statistics')
display(rfm[['R','F','M']].describe().round(2))
````

````python
#RFM Scoring: quintile-based 1–5 scoring
# Recency:  LOWER recency_days = better → inverted scoring
# Frequency: HIGHER = better
# Monetary:  HIGHER = better

rfm['R_score'] = pd.qcut(rfm['R'], q=5, labels=[5, 4, 3, 2, 1]).astype(int)
rfm['F_score'] = pd.qcut(rfm['F'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5]).astype(int)
rfm['M_score'] = pd.qcut(rfm['M'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5]).astype(int)

# Composite RFM score (simple sum)
rfm['RFM_Score']     = rfm['R_score'] + rfm['F_score'] + rfm['M_score']

# RFM string code (e.g. '545' = high recency, high frequency, high monetary)
rfm['RFM_Code'] = rfm['R_score'].astype(str) + rfm['F_score'].astype(str) + rfm['M_score'].astype(str)

print('RFM Scores Added:')
display(rfm.head(8))
print(f'\nRFM Score Range: {rfm["RFM_Score"].min()} – {rfm["RFM_Score"].max()}')
````

## 9. RFM Scoring & Segmentation

Using the RFM composite score, we assign each customer to one of **five named business segments**. These segments are designed to be immediately actionable by a marketing team.

````python
#Segment assignment logic 
def assign_rfm_segment(row):
    r, f, m = row['R_score'], row['F_score'], row['M_score']
    score   = row['RFM_Score']

    # Champions: high across all three dimensions
    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'
    # Loyal Customers: high frequency and monetary, moderate recency
    elif f >= 4 and m >= 4:
        return 'Loyal Customers'
    # At Risk: previously active but recently disengaged
    elif r <= 2 and f >= 3:
        return 'At Risk'
    # New Customers: recent signup, low frequency
    elif r >= 4 and f <= 2:
        return 'New Customers'
    # Hibernating: low recency and low frequency
    elif score <= 6:
        return 'Hibernating'
    else:
        return 'Potential Loyalists'

rfm['RFM_Segment'] = rfm.apply(assign_rfm_segment, axis=1)

# Merge segment back to main dataframe
df = df.merge(rfm[['customer_id', 'R_score', 'F_score', 'M_score', 'RFM_Score', 'RFM_Code', 'RFM_Segment']],
              on='customer_id', how='left')

print('RFM Segment Distribution:')
seg_counts = rfm['RFM_Segment'].value_counts()
display(pd.DataFrame({'Segment': seg_counts.index,
                      'Count': seg_counts.values,
                      'Share (%)': (seg_counts.values / len(rfm) * 100).round(1)}))
````

## 10. K-Means Clustering

While RFM segmentation applies rule-based logic, **K-Means Clustering** discovers natural groupings in the data without pre-defined rules. This allows us to validate the RFM segments and potentially surface patterns that manual scoring misses.

### 10.1  Feature Preparation

````python
#Select clustering features 
# We use the core behavioral features that define customer value
cluster_features = ['recency_days', 'frequency', 'monetary_value',
                    'email_open_rate', 'churn_risk_score']

X = df[cluster_features].copy()

print('Clustering features selected:')
for f in cluster_features:
    print(f'  • {f}')

# Verify no missing values
print(f'\nMissing values in feature matrix: {X.isnull().sum().sum()}')
print(f'Feature matrix shape: {X.shape}')
````

````python
#Feature scaling — StandardScaler
# K-Means is distance-based → features MUST be on the same scale

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=cluster_features)

print('Feature scaling complete (StandardScaler)')
print('\nScaled feature statistics:')
display(X_scaled_df.describe().round(3))
````

### 10.2  Optimal k: Elbow Method + Silhouette Score

````python
#Finding optimal k
k_range      = range(2, 11)
inertia_vals = []
sil_vals     = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertia_vals.append(km.inertia_)
    sil_vals.append(silhouette_score(X_scaled, labels))
    print(f'  k={k} | Inertia: {km.inertia_:,.1f} | Silhouette: {silhouette_score(X_scaled, labels):.4f}')
````

````python
#Elbow + Silhouette visualisation 
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
fig.suptitle('Optimal Number of Clusters (k)', fontsize=15, fontweight='bold', y=1.02)

# Elbow method
axes[0].plot(list(k_range), inertia_vals, 'o-', color=NAVY, linewidth=2.5, markersize=7)
axes[0].fill_between(list(k_range), inertia_vals, alpha=0.08, color=SKY)
axes[0].set_xlabel('Number of Clusters (k)')
axes[0].set_ylabel('Inertia (Within-Cluster Sum of Squares)')
axes[0].set_title('Elbow Method')
axes[0].set_xticks(list(k_range))

# Silhouette score
best_k_idx = sil_vals.index(max(sil_vals))
bar_colors = [NAVY if i == best_k_idx else SKY for i in range(len(k_range))]
axes[1].bar(list(k_range), sil_vals, color=bar_colors, edgecolor='white', alpha=0.85)
axes[1].set_xlabel('Number of Clusters (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Score by k')
axes[1].set_xticks(list(k_range))
axes[1].text(list(k_range)[best_k_idx], max(sil_vals) + 0.002,
             f'Best k={list(k_range)[best_k_idx]}', ha='center', fontweight='bold', color=NAVY)

plt.tight_layout()

plt.savefig(
    "../visuals/charts/elbow_method.png",
    bbox_inches="tight",
    dpi=300
)

plt.show()

OPTIMAL_K = list(k_range)[best_k_idx]
print(f'\n✅ Optimal k = {OPTIMAL_K}  |  Silhouette Score = {max(sil_vals):.4f}')
````

### 10.3  Fit Final K-Means Model

````python
#Train final model 
kmeans_final = KMeans(
    n_clusters  = OPTIMAL_K,
    random_state= 42,
    n_init      = 20,       # Multiple initialisations for stability
    max_iter    = 500
)

df['Cluster'] = kmeans_final.fit_predict(X_scaled)

final_sil = silhouette_score(X_scaled, df['Cluster'])

print('━' * 50)
print(f'  K-MEANS MODEL TRAINED')
print('━' * 50)
print(f'  Clusters    : {OPTIMAL_K}')
print(f'  Inertia     : {kmeans_final.inertia_:,.2f}')
print(f'  Silhouette  : {final_sil:.4f}')
print('━' * 50)
print(f'\nCluster distribution:')
print(df['Cluster'].value_counts().sort_index())
````

### 10.4  Cluster Visualisation via PCA

````python
#Reduce to 2D for visualisation using PCA
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

print(f'PCA Variance Explained: PC1={pca.explained_variance_ratio_[0]:.1%}, '
      f'PC2={pca.explained_variance_ratio_[1]:.1%} '
      f'(total: {sum(pca.explained_variance_ratio_):.1%})')

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('K-Means Cluster Visualisation (PCA 2D Projection)', fontsize=15, fontweight='bold', y=1.02)

cluster_palette = [NAVY, SKY, '#e8a030', '#2a9d8f', '#e76f51', '#8338ec', '#6c757d']

for c in sorted(df['Cluster'].unique()):
    mask = df['Cluster'] == c
    axes[0].scatter(X_pca[mask, 0], X_pca[mask, 1],
                    c=cluster_palette[c], label=f'Cluster {c}',
                    alpha=0.6, s=25, edgecolors='none')

# Plot centroids on PCA space
centers_pca = pca.transform(kmeans_final.cluster_centers_)
axes[0].scatter(centers_pca[:, 0], centers_pca[:, 1],
                c='black', marker='X', s=200, zorder=5, label='Centroids')
axes[0].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.1%} variance)')
axes[0].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.1%} variance)')
axes[0].set_title('Cluster Separation (PCA)')
axes[0].legend(fontsize=9)

# Cluster size chart
cluster_counts = df['Cluster'].value_counts().sort_index()
axes[1].bar([f'Cluster {c}' for c in cluster_counts.index],
            cluster_counts.values,
            color=[cluster_palette[c] for c in cluster_counts.index],
            edgecolor='white')
axes[1].set_ylabel('Number of Customers')
axes[1].set_title('Cluster Sizes')
for i, v in enumerate(cluster_counts.values):
    axes[1].text(i, v + 2, str(v), ha='center', fontsize=11, fontweight='bold')

plt.tight_layout()

plt.savefig(
    "../visuals/charts/customer_clusters.png",
    bbox_inches="tight",
    dpi=300
)

plt.show()
````

Long inactive, low engagement. **Cost of reactivation likely exceeds LTV for most.** Last-chance offer for a small subset; sunset the rest to protect email deliverability.
